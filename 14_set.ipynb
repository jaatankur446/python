# SET — Data Engineering Scenarios

Set exercises for reconciliation, schema comparison, duplicate removal, and cross-system matching.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on SET.


## Scenario 1

Find customer IDs common to CRM and billing.


````python
crm={101,102,103,104}; billing={103,104,105}
print(sorted(crm&billing))
````

### Expected Output
```text
[103, 104]
```


## Scenario 2

Find source IDs missing from the warehouse.


````python
source={1,2,3,4,5}; warehouse={1,3,5}
print(sorted(source-warehouse))
````

```text

```
