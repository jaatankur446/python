## Scenario 1

Create a PipelineJob class with name and status.


````python
class PipelineJob:
    def __init__(self,name):
        self.name=name; self.status="CREATED"
job=PipelineJob("customer_load")
print(job.name,job.status)
````

### Expected Output
```text

```


## Scenario 2

Add a start method to a pipeline class.


````python
class Job:
    def __init__(self,name): self.name=name; self.status="CREATED"
    def start(self): self.status="RUNNING"
job=Job("orders")
job.start()
print(job.status)
````

```text
RUNNING
```

## Scenario 15

Demonstrate method overriding for file formats.


````python
class Writer:
    def write(self): print("generic")
class CSVWriter(Writer):
    def write(self): print("csv")
class JSONWriter(Writer):
    def write(self): print("json")
for w in [CSVWriter(),JSONWriter()]: w.write()
````

### Expected Output
```text
csv
json
```

## Scenario 17

Create a method to calculate a partition's density.


````python
class Partition:
    def __init__(self,rows,size_mb): self.rows=rows; self.size_mb=size_mb
    def density(self): return self.rows/self.size_mb
print(Partition(20000,100).density())
````

## Scenario 19

Create an ETL class with extract, transform, load methods.


````python
class ETL:
    def extract(self): return [1,2,3]
    def transform(self,data): return [x*10 for x in data]
    def load(self,data): print(data)
job=ETL(); job.load(job.transform(job.extract()))
````

### Expected Output
```text
[10, 20, 30]
```

## Scenario 22

Use polymorphism with different loaders.


````python
class Loader:
    def load(self): pass
class DBLoader(Loader):
    def load(self): print("Database")
class FileLoader(Loader):
    def load(self): print("File")
for x in [DBLoader(),FileLoader()]: x.load()
````

### Expected Output
```text
Database
File
```

## Scenario 27

Create a class that checks whether a job can retry.


````python
class RetryPolicy:
    def can_retry(self,status,retries): return status=="FAILED" and retries<3
print(RetryPolicy().can_retry("FAILED",2))
````

### Expected Output
```text

```
