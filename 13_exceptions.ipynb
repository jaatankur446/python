## Scenario 1

Handle an invalid integer in a source feed.


````python
values=["100","200","N/A","300"]
for x in values:
    try: print(int(x))
    except ValueError: print("INVALID",x)
````

### Expected Output
```text
100
200
INVALID N/A
300
```

## Scenario 16

Handle a malformed key-value line.


````python
lines=["host=db1","port=3306","invalid"]
for line in lines:
    try:
        key,value=line.split("=")
        print(key,value)
    except ValueError:
        print("Malformed line:",line)
````

### Expected Output
```text
host db1
port 3306
Malformed line: invalid
```


## Scenario 17

Create a custom exception for invalid batch size.


````python
class InvalidBatchSize(Exception): pass
def check(size):
    if size<=0: raise InvalidBatchSize("Batch size must be positive")
    return "OK"
try: print(check(0))
except InvalidBatchSize as e: print(e)
````

### Expected Output
```text
Batch size must be positive
```


## Scenario 18

Handle API timeout as a custom error simulation.


````python
class APITimeout(Exception): pass
try:
    raise APITimeout("API request timed out")
except APITimeout as e:
    print(e)
````

### Expected Output
```text
API request timed out
```

## Scenario 20

Prevent one bad row from stopping a loop.


````python
rows=["10","20","bad","40"]
total=0
for row in rows:
    try: total+=int(row)
    except ValueError: continue
print(total)
````

### Expected Output
```text
70
```

## Scenario 29

Continue processing after a failed database conversion.


````python
values=["101","102","x","104"]
processed=[]
for x in values:
    try: processed.append(int(x))
    except ValueError: print("Skipped",x)
print(processed)
````

### Expected Output
```text

[101, 102, 104]
```
