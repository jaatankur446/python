# Dictionary — Data Engineering Scenarios

Dictionary exercises for aggregation, lookups, metadata, configuration, and pipeline status.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Dictionary.


## Scenario 1

Aggregate sales by city using a dictionary.


````python
orders=[("Delhi",10000),("Mumbai",15000),("Delhi",5000)]
totals={}
for city,amount in orders: totals[city]=totals.get(city,0)+amount
print(totals)
````

### Expected Output
```text
{'Delhi': 15000, 'Mumbai': 15000}
```


## Scenario 2

Count data-quality statuses.


````python
statuses=["VALID","FAILED","VALID","WARNING","FAILED"]
counts={}
for s in statuses: counts[s]=counts.get(s,0)+1
print(counts)
````

### Expected Output
```text
{'VALID': 2, 'FAILED': 2, 'WARNING': 1}
```


## Scenario 3

Create a customer lookup dictionary.


````python
customers=[(101,"Rahul"),(102,"Priya"),(103,"Amit")]
lookup={cid:name for cid,name in customers}
print(lookup[102])
````

### Expected Output
```text
Priya
```


## Scenario 4

````python
metrics={"rows":12000,"errors":25,"warnings":80}
print(max(metrics,key=metrics.get))
````

## Scenario 6

Filter dictionary entries with values above a threshold.


````python
sales={"Delhi":50000,"Mumbai":75000,"Patna":30000}
print({k:v for k,v in sales.items() if v>50000})
````


## Scenario 10

Merge two configuration dictionaries.


````python
a={"host":"db1","port":3306}
b={"database":"sales","user":"etl"}
config={**a,**b}
print(config)
````

### Expected Output
```text
{'host': 'db1', 'port': 3306, 'database': 'sales', 'user': 'etl'}
```
