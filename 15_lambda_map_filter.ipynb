# Lambda, Map and Filter — Data Engineering Scenarios

Functional-programming exercises for filtering, mapping, sorting, and lightweight data transformations.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Lambda, Map and Filter.


## Scenario 1

Sort files by size using lambda.


````python
files=[("a.csv",50),("b.csv",20),("c.csv",80)]
print(sorted(files,key=lambda x:x[1]))
````

### Expected Output
```text
[('b.csv', 20), ('a.csv', 50), ('c.csv', 80)]
```


## Scenario 2

Filter transactions above 10000.


````python
values=[5000,12000,8000,25000]
print(list(filter(lambda x:x>10000,values)))
````

## Scenario 30

Map and filter valid transaction amounts in one pipeline.


````python
values=["5000","12000","bad","25000"]
valid=filter(lambda x:x.isdigit() and int(x)>10000,values)
print(list(map(int,valid)))
````

### Expected Output
```text
[12000, 25000]
```
