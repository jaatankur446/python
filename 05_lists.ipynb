# Lists — Data Engineering Scenarios

List-processing exercises for records, partitions, transactions, queues, and transformations.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Lists.


## Scenario 1

Remove duplicate customer IDs while preserving order.


````python
ids=[101,102,101,103,102,104]
result=[]
for x in ids:
    if x not in result: result.append(x)
print(result)
````

### Expected Output
```text
[101, 102, 103, 104]
```

## Scenario 13

Calculate cumulative totals.


````python
amounts=[100,200,150]
total=0
result=[]
for x in amounts:
    total+=x; result.append(total)
print(result)
````

### Expected Output
```text
[100, 300, 450]
```

## Scenario 20
