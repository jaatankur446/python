## Scenario 1

A customer record should be accepted only when age is 18–70 and status is ACTIVE. Classify each record.


````python
records=[(101,25,"ACTIVE"),(102,17,"ACTIVE"),(103,42,"INACTIVE"),(104,71,"ACTIVE")]
for cid,age,status in records:
    if 18<=age<=70 and status=="ACTIVE": result="VALID"
    elif 18<=age<=70: result="REVIEW"
    else: result="REJECT"
    print(cid,result)
````

### Expected Output
```text
101 VALID
102 REJECT
103 REVIEW
104 REJECT
```

## Scenario 29

Determine whether a batch meets a minimum success rate of 95%.


````python
batches=[("B1",980,1000),("B2",930,1000),("B3",950,1000)]
for bid,success,total in batches:
    rate=success/total*100
    print(bid,"PASS" if rate>=95 else "FAIL")
````
