# Strings — Data Engineering Scenarios

String processing exercises based on logs, source data, identifiers, file names, and pipeline metadata.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Strings.


## Scenario 1

Normalize source-system names by trimming spaces and converting to title case.


````python
names=["  rahul kumar ","PRIYA SINGH"," amit  sharma "]
for name in names:
    print(" ".join(name.strip().title().split()))
````

### Expected Output
```text
Rahul Kumar
Priya Singh
Amit Sharma
```


## Scenario 2

Extract the domain from an email address.


````python
emails=["a@gmail.com","b@company.com","c@data.org"]
for email in emails:
    print(email.split("@")[1])
````

### Expected Output
```text
gmail.com
company.com
data.org
```

## Scenario 14

Count vowels in data labels.


````python
labels=["customer","transaction","warehouse"]
for label in labels:
    print(sum(1 for c in label.lower() if c in "aeiou"))
````

### Expected Output
```text
3
4
4
```

## Scenario 30

Generate a partition path using year and month strings.
