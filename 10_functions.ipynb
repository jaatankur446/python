## Scenario 1

Create a function to calculate success rate.


````python
def success_rate(total,success):
    return 0 if total==0 else success/total*100
print(success_rate(1000,970))
````

### Expected Output
```text
97.0
```

Create a function to calculate average file size.


````python
def average_size(sizes):
    return sum(sizes)/len(sizes)
print(average_size([10,20,30]))
````

## Scenario 12

Create a function to remove duplicates while preserving order.


````python
def unique(values):
    result=[]
    for x in values:
        if x not in result: result.append(x)
    return result
print(unique([1,2,1,3,2]))
````

### Expected Output
```text
[1, 2, 3]
```

## Scenario 24

Create a function to calculate running totals.


````python
def running_total(values):
    total=0; result=[]
    for x in values:
        total+=x; result.append(total)
    return result
print(running_total([100,200,50]))
````

### Expected Output
```text
[100, 300, 350]
```


## Scenario 25

Create a function to check whether an email has a basic valid format.


````python
import re
def valid_email(email):
    return re.fullmatch(r"[^@\s]+@[^@\s]+\.[A-Za-z]{2,}",email) is not None
print(valid_email("a@example.com"))
````

### Expected Output
```text
True
```


## Scenario 26

Create a function to convert a date string to a partition key.


````python
from datetime import datetime
def partition_key(value):
    return datetime.strptime(value,"%Y-%m-%d").strftime("%Y%m%d")
print(partition_key("2026-10-06"))
````

### Expected Output
```text
20261006
```

## Scenario 28

Create a function to return the smallest positive value.


````python
def smallest_positive(values):
    positives=[x for x in values if x>0]
    return min(positives) if positives else None
print(smallest_positive([-5,0,12,4]))
````

### Expected Output



## Scenario 29

Create a function to build a simple record dictionary.


````python
def make_record(schema,values):
    return dict(zip(schema,values))
print(make_record(["id","city"],[101,"Delhi"]))
````

### Expected Output
```text

```


## Scenario 30

Create a function to summarize pipeline status counts.


````python
def status_summary(statuses):
    result={}
    for s in statuses: result[s]=result.get(s,0)+1
    return result
print(status_summary(["SUCCESS","FAILED","SUCCESS"]))
````

### Expected Output
```text
{'SUCCESS': 2, 'FAILED': 1}
```
