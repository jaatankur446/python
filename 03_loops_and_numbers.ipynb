## Scenario 1

A pipeline receives numbers. Print all even values.


````python
values=[12,18,25,31,44,50]

evens=[]
for n in values:
    if n%2==0: evens.append(n)
print(evens)
````

## Scenario 11

Find the second largest number in a list.


````python
values=[15,42,28,71,56]

largest=second=float('-inf')
for n in values:
    if n>largest:
        second=largest; largest=n
    elif largest>n>second:
        second=n
print(second)
````

## Scenario 17

Find all duplicate numeric IDs using loops.


````python
values=[101,102,101,103,104,102]

duplicates=[]
for n in values:
    if values.count(n)>1 and n not in duplicates: duplicates.append(n)
print(duplicates)
````
