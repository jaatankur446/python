# Regex — Data Engineering Scenarios

Regular-expression exercises for log parsing, validation, extraction, masking, and data cleaning.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Regex.


## Scenario 1

Extract email addresses from a log message.


````python
import re
text="Contact ops@example.com or data.team@company.org for support."
print(re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",text))
````

### Expected Output
```text
['ops@example.com', 'data.team@company.org']
```


## Scenario 2

Extract IPv4 addresses from server logs.


````python
import re
text="src=192.168.1.10 dst=10.20.30.40"
print(re.findall(r"\b(?:\d{1,3}\.){3}\d{1,3}\b",text))
````

### Expected Output
```text
['192.168.1.10', '10.20.30.40']
```


## Scenario 3

Validate a simple email field.


````python
import re
values=["a@example.com","bad-email","x@y.org"]
pattern=r"^[\w.-]+@[\w.-]+\.[A-Za-z]{2,}$"
for x in values: print(x,bool(re.match(pattern,x)))
````

### Expected Output
```text
a@example.com True
bad-email False
x@y.org True
```

## Scenario 10

Mask phone numbers while retaining the last four digits.


````python
import re
phones=["9876543210","9123456789"]
for p in phones: print(re.sub(r"\d(?=\d{4})","X",p))
````

### Expected Output
```text

XXXXXX6789
```

## Scenario 12

Extract file extensions.

## Scenario 14

Extract table names from simple SQL statements.

## Scenario 23

Find repeated words in a simple text.


````python
import re
text="load failed load retry"
words=re.findall(r"\b\w+\b",text.lower())
print([w for w in set(words) if words.count(w)>1])
````

### Expected Output

## Scenario 29

Find log lines containing WARNING or ERROR.


````python
import re
text="INFO ok\nWARNING slow\nERROR failed\nINFO done"
print(re.findall(r"^(?:WARNING|ERROR).*$",text,re.M))
````

### Expected Output
```text
['WARNING slow', 'ERROR failed']
```
