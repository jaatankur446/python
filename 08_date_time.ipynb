## Scenario 1

Calculate data freshness in minutes.


````python
from datetime import datetime
loaded=datetime.strptime("2026-10-06 10:15:00","%Y-%m-%d %H:%M:%S")
now=datetime.strptime("2026-10-06 11:05:00","%Y-%m-%d %H:%M:%S")
print(int((now-loaded).total_seconds()/60))
````

### Expected Output
```text
50
```

## Scenario 3

## Scenario 13

Filter records from the last 30 days relative to a reference date.


````python
from datetime import datetime,timedelta
ref=datetime(2026,10,30)
values=[datetime(2026,10,5),datetime(2026,10,15),datetime(2026,9,20)]
for dt in values:
    if ref-timedelta(days=30)<=dt<=ref: print(dt.strftime("%Y-%m-%d"))
````

```text

```

## Scenario 21

Group timestamps by month.


````python
from datetime import datetime
values=["2026-01-10","2026-01-20","2026-02-05"]
counts={}
for x in values:
    m=datetime.strptime(x,"%Y-%m-%d").strftime("%Y-%m")
    counts[m]=counts.get(m,0)+1
print(counts)
````

### Expected Output
```text
{'2026-01': 2, '2026-02': 1}
```


## Scenario 22

Find the processing date after three business days for a simple weekday example.


````python
from datetime import datetime,timedelta
dt=datetime(2026,10,5)
added=0
while added<3:
    dt+=timedelta(days=1)
    if dt.weekday()<5: added+=1
print(dt.strftime("%Y-%m-%d"))
````

### Expected Output
```text
2026-10-08
```
