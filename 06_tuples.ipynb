# Tuples — Data Engineering Scenarios

Tuple exercises using schemas, API responses, immutable metadata, and record structures.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Tuples.


## Scenario 1

Store schema columns as tuples and print each column.


````python
schema=(("id","INT"),("name","STRING"),("amount","DECIMAL"))
for col,dtype in schema: print(col,dtype)
````

### Expected Output
```text
id INT
name STRING
amount DECIMAL
```


## Scenario 2

Find the endpoint with maximum response time.


````python
responses=(("/users",120),("/orders",320),("/payments",180))
print(max(responses,key=lambda x:x[1]))
````

### Expected Output
```text
('/orders', 320)
```


## Scenario 3

## Scenario 20

## Scenario 30

Find the total of all partition counts stored as tuple records.


````python
partitions=(("p01",1200),("p02",1800),("p03",1500))
total=0
for name,count in partitions:
    total += count
print(total)
````

### Expected Output
```text

```
