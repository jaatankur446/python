# Files I/O — Data Engineering Scenarios

File-processing exercises for logs, CSV-style data, configuration, reports, and rejected records.


## Contents

This notebook contains **30 practical Data Engineering scenarios** focused on Files I/O.


## Scenario 1

Write a daily pipeline status to a text file.


````python
with open("daily_status.txt","w") as f:
    f.write("customer_load: SUCCESS\n")
with open("daily_status.txt") as f:
    print(f.read(),end="")
````

### Expected Output
```text
customer_load: SUCCESS
```


## Scenario 2

Read a CSV-like file and count data rows.


````python
data="id,city\n1,Delhi\n2,Mumbai\n3,Patna\n"
with open("orders.csv","w") as f: f.write(data)
with open("orders.csv") as f: print("Rows:",len(f.readlines())-1)
````

### Expected Output
```text
Rows: 3
```

## Scenario 4

Create a rejected-record file.


````python
rejected=[101,105,110]
with open("rejected.txt","w") as f:
    for x in rejected: f.write(str(x)+"\n")
with open("rejected.txt") as f: print(f.read(),end="")
````

### Expected Output
```text
101
105
110
```


## Scenario 5

Count lines containing ERROR.


````python
text="INFO start\nERROR timeout\nINFO retry\nERROR missing file\n"
with open("app.log","w") as f: f.write(text)
with open("app.log") as f:
    print(sum(1 for line in f if "ERROR" in line))
````

### Expected Output
```text
2
```


## Scenario 6

Read a file and remove blank lines.


````python
text="101\n\n102\n\n103\n"
with open("ids.txt","w") as f: f.write(text)
with open("ids.txt") as f:
    print([line.strip() for line in f if line.strip()])
````

### Expected Output
```text
['101', '102', '103']
```

## Scenario 8

Copy selected records from one file to another.


````python
source="101\n102\n103\n"
with open("source.txt","w") as f: f.write(source)
with open("source.txt") as src, open("selected.txt","w") as dst:
    for line in src:
        if int(line.strip())%2==1: dst.write(line)
with open("selected.txt") as f: print(f.read(),end="")
````

### Expected Output
```text
101
103
```

## Scenario 13

Write a summary report.


````python
summary={"success":950,"failed":30}
with open("summary.txt","w") as f:
    for k,v in summary.items(): f.write(f"{k}={v}\n")
with open("summary.txt") as f: print(f.read(),end="")
````

### Expected Output

## Scenario 14

Read only the first two records from a file.


````python
with open("data.txt","w") as f: f.write("A\nB\nC\n")
with open("data.txt") as f:
    for i,line in enumerate(f):
        if i==2: break
        print(line.strip())
````

### Expected Output
```text
A
B
```


## Scenario 15

Create a file containing failed job names.


````python
jobs=["load1","load2","load3"]
failed=["load2"]
with open("failed_jobs.txt","w") as f:
    for j in jobs:
        if j in failed: f.write(j+"\n")
with open("failed_jobs.txt") as f: print(f.read(),end="")
````

### Expected Output
```text
load2
```

## Scenario 19

Read JSON-like simple lines as key-value pairs.


````python
with open("config.txt","w") as f: f.write("host=db01\nport=3306\n")
config={}
with open("config.txt") as f:
    for line in f:
        k,v=line.strip().split("="); config[k]=v
print(config)
````

### Expected Output
```text
{'host': 'db01', 'port': '3306'}
```

## Scenario 21

Create a file with only high-value transactions.


````python
tx=[("T1",5000),("T2",25000),("T3",12000)]
with open("high_value.txt","w") as f:
    for tid,amount in tx:
        if amount>10000: f.write(f"{tid},{amount}\n")
with open("high_value.txt") as f: print(f.read(),end="")
````

### Expected Output
```text
T2,25000
T3,12000
```
